```python
import csv
from bs4 import BeautifulSoup
```

```python
# Open and read the html file (remember that your file path may be different!!!)
with open("data/project-gutenberg.html", "r") as page:

    # Tell the BeautifulSoup library to interpret the file
    soup = BeautifulSoup(page, "html.parser")

    # Find the first h1 on the page. Print the text stored within it.
    h1 = soup.find("h1").text
    print(h1)
    
    # There are three follow links on the page, one for Facebook, one for Bluesky, and one for Mastodon. 
    # What is the class associated with these follow links? Answer here: "grayed navlink"
    # What html element are these follow links stored in? Answer here: <li class="grayed navlink"> <li>

    # Use the function soup.find_all(...) to grab the html associated with the three follow links and store them in an array. 
    sm_list_items = soup.find_all("li", {"class":"grayed navlink"})
    ## print(sm_links)

    # Loop through the array, printing the text for each follow link.
    for list_item in sm_list_items:
        list_item_text = list_item.text
        ## print(list_item_text)

    # Let's take it a step further. Paste your loop again below. In each iteration of the loop, grab the first <span></span> that you find 
    # in the html code and print out its text.
    for list_item in sm_list_items:
        list_item_title = list_item.find("span",{"class":"title"}).text
        print(list_item_title)

```
